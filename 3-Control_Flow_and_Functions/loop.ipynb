# Loops in Python

Loops are used to iterate over sequences (like lists, tuples, or strings) or execute a block of code repeatedly as long as a condition is true.

There are two primary types of loops in Python:

- `for` loops, which iterate over a sequence.
- `while` loops, which continue to execute as long as a given condition is `True`.

```python
# 1. Looping through data structures

# for loop
numbers = [1, 2, 3, 4, 5]
for number in numbers:
    print(number)

# while loop
count = 0
while count < 5:
    print(f"Count is: {count}")
    count += 1
```

```python
# 2. Examples: summation, factorial, and other repetitive calculations

# Summation of numbers
total_sum = 0
for number in range(1, 101):
    total_sum += number
print(total_sum)

# Factoriral calculation
number = int(input("Enter a number: "))
factorial = 1
while number > 0:
    factorial *= number
    number -= 1
print(factorial)
```
