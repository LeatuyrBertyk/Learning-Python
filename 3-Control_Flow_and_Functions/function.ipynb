# Function in Python

Functions are blocks of reusable code that perform a specific task. They help to break programs into smaller, modular pieces, making the code mode readable and maintainable.

To define a function in Python, you can the `def` keyword, followed by the function name and parentheses. Inside the parenthesess, you can specify paremeters (if any), and the function body follows under an indented block.

```python
# 1. How to define functions in Python

def function_name(parameters):
    # Function body
    value = 1
    return value

# For example, define a function that takes two numbers and returns their sum:
def add_numbers(a, b):
    return a + b

# Call the function by passing the appropriate arguments:
result = add_numbers(3, 4)
print(result)
```

```python

# 2. Lambda fuctions (anonymous functions) for quick calculations

# Lambda functions, also known as anonymous functions, are small functions without a name. They can take any number of arguments but have only one expression. You define them using keyword `lambda`.

multiply = lambda x, y: x * y
print(multiply(4,5))
```

```python
# 3. Using `yield` to optimize memory

# The system do not use any memory block to store variable, it temporarily break.

def generate(n):
    for i in range(1, n):
        yield i

my_gen = generate(5)
ans = ""
for num in my_gen:
    ans += f"{str(num)} "
print(ans)

# Other method to access values
# print(next(my_gen))
# print(next(my_gen))
# print(next(my_gen))
# print(next(my_gen))
# print(next(my_gen))
```
