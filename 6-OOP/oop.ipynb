# Python OOP

## What is OOP?

**OOP** stands for **Object-Oriented Programming**. Python is an object-oriented language, allowing you to structure your code using classes and objects for better organization and reusability.

## Advantages of OOP

- Provides a clear structure to programs
- Makes code easier to maintain, reuse, and debug
- Helps keep your code DRY (**Don't repeat yourself**)
- Allows you to build reusable applications with less code

**Tip**: The DRY principle  means you should avoid writting the same code more than once. Move repeated code into functions or classes and reuse it.

## OOP properties

OOP has 3 main properties: Inhertance, Polymorphism, and Encapsulation.

```python
# Class Person
class Person:
    def __init__(self, name, age, sex):
        self.name = name
        self.age = age
        self.sex = sex

    def __str__(self):
        return f"{'He' if self.sex == 'male' else 'She'} is {self.name}, {self.age} years old"

    def fuck(self):
        print(f"A a aa...")

# In this class: name, age, sex are call property; __init__, __str__, fuck are called method

# __init__: object initialization
# __str__: determine how the class is printed when called

p1 = Person("Tom", 20, "male")
print(p1)
print(p1.name)
print(p1.age)

## Property 1: Inheritance
class Children(Person):
    def __init__(self, name, age, sex, school):
        super().__init__(name, age, sex)
        self.school = school

    def __str__(self):
        return f"{super().__str__()} and study at {self.school} school"
    
    def work(self):
        print("Studying")

c1 = Children("Anna", 6, "male", "CVA")
print(c1)
c1.work()

## Property 2: Polymorphism
class Teenager(Children):
    def work(self):
        print("Studying but more subjects")

t1 = Teenager("Anna", "18", "female", "HLK")
t1.work()

## Property 3: Encapsulation
class User:
    def __init__(self, username, password):
        self.username = username
        self.__password = password # Private property
    def get_password(self): # Method to get a private property
        return self.__password

u1 = User("Leatuyr Bertyk", "123456")
print(u1.username)
# print(u1.__password) ## Error
print(u1.get_password())
```

## Public vs. Protected vs. Private

```python
class MyClass:
    def __init__(self, public, protected, private):
        self.public = public        # Public
        self._protected = protected # Proteced
        self.__private = private    # Private

obj = MyClass("Public", "Protected", "Private")
print(obj.public) # Can be accessed
print(obj._protected) # Can be accessed, but shouldn't
# print(obj.__private) # Error: Cannot be accessed

# We are able to access or modify protected property, but this is not encouraged as it is a convention. 
# This property is used for internal method, should be accessed by only other class or inner class, by inherentance.
```

## Delete property and object

```python
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year
    def run(self):
        return f"{self.brand}-{self.model} was produced in {self.year}"

c1 = Car("Ford", "Mustang", "2025")
print(c1.run())

del c1.brand
# print(c1.brand) # Error: c1.brand is deleted

del c1
# print(c1.brand()) # Error: c1 is deleted
```
