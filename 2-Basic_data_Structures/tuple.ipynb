# Tuple in Python

Tuples are similar to lists in Python, but with one key difference: tuples are immutable, meaning that once a tuple is created, its elements cannnot be modified.

If you can to modify a tuple, you can create a new list based on this tuple, then convert this list to tuple after modifying.

Tuples are defined using parentheses `()`.

```python
# 1. Introduction to Tuples: Immutable sequences

my_tuple = (1, 2, 3)
print(my_tuple)

my_tuple = 4, 5, 6
print(my_tuple)
```

```python
# 2. Tuple unpacking and basic operations

# Tuple unpacking allows you to assign the values of ta tuple to multiple variables in one step.

coordinates = (10, 20)
(x, y) = coordinates

print(x)
print(y)

# Although tuples are immutable, you can perform other operations like indexing and slicing, similar to lists.
my_tuple = (5, 10, 15, 20)
print(my_tuple[1])
print(my_tuple[:3])

# Modify tuple:
temp_list = list(coordinates)
temp_list.append(30)
coordinates = tuple(temp_list)

(x, y, z) = coordinates
print(f"{x} {y} {z}")

# Unpacking tuple:
my_tuple = (10, 20, 30, 40, 50)
(x, *y, z, t) = my_tuple
print(x)
print(y)
print(z)
print(t)
```
