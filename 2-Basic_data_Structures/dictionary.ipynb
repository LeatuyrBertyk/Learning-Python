# Dictionary in Python

A dictionary is a collection of key-value pairs, where each key is associated with a value. Dictionaries are unordered, mutable, and the keys must be unique. Dictionaries are defined using curly braces `{}`.

Dictionaries are similar to maps in `C/C++`, and have the same syntax as `JSON`.

```python
# 1. Key-Value pair data structure

my_dict = {"name": "Alice", "age": 25, "city": "New York"}
print(my_dict)
```

```python
# 2. Common dictionary operations (access, update, remove)

# To access the value associated with a key, you can use the key in square brackets[], you can also use the .get() method.
print(my_dict["name"])
print(my_dict.get("age"))

# To add or update a key-value pair, you can simply assign a vallue to the key
my_dict["age"] = 26
my_dict["email"] = "alice@example.com"
print(my_dict)

# To remove a key-value pair, you can use the pop() method.
my_dict.pop("city")
```

```python
# 3. Dictionary and Json

import json

# Store json to a variable as a dictionary 
car = '{"brand": "Ford", "model": "Mustang", "year": 2020}'
x = json.loads(car)

print(x)

# Get json from a dictionary
y = json.dumps(x)
print(y)
```
