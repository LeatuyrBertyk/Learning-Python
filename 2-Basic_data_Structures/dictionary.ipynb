# Dictionary in Python

A dictionary is a collection of key-value pairs, where each key is associated with a value. Dictionaries are unordered, mutable, and the keys must be unique. Dictionaries are defined using curly braces `{}`.

Dictionaries are similar to maps in `C/C++`, and have the same syntax as `JSON`.

## Key-value pair data structure

```python
my_dict = {"name": "Alice", "age": 25, "city": "New York"}
print(my_dict)
```

## Common dictionary operations: Access, Update, and Remove

```python
my_dict = {"name": "Alice", "age": 25, "city": "New York"}

# 1. To access the value associated with a key, you can use the key in square brackets[], you can also use the .get() method:
print(my_dict["name"])    # Method 1
print(my_dict.get("age")) # Method 2

# Access keys:
for key in my_dict.keys():
    print(key)

# Access values
for value in my_dict.values():
    print(value)

# Access key-value pairs:
for key, value in my_dict.items():
    print(f"{key}: {value}")
    

# 2. To add or update a key-value pair, you can simply assign a vallue to the key:
my_dict["age"] = 26
my_dict["email"] = "alice@example.com"
print(my_dict)

# 3. To remove a key-value pair, you can use the pop() method:
city_value = my_dict.pop("city")
```

## Dictionary and JSON

```python
import json

# Store json to a variable as a dictionary 
car = '{"brand": "Ford", "model": "Mustang", "year": 2020}'
x = json.loads(car)

print(x)

# Get json from a dictionary
y = json.dumps(x)
print(y)
```
