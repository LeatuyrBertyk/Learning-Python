# List in Python

A list is one of the most commonly used data structures in Python. A list is a collection of items that can hold different types of elements, such as integers, floats, strings, or even other lists. Lists in Python are ordered and mutable, meaning that their elements can be modified after the list is created.

To define a list in Python, you can use square brackets `[]` and seperate the elements with commas.

```python
# 1. Defining Lists and Basic operations

# Defining a simple list
my_list = [1, 2, 3, 4, 5]
print(my_list)

# List with different data types
mixed_list = [1, "Hello", 3.14, True]
print(mixed_list)
```

```python
# 2. Indexing and Slicing in Lists

# The index of the first element in a list is 0, the second element is at index 1, so on.
# Negative indexing starts from the last element with index -1

my_list = [10, 20, 30, 40, 50]
print(my_list[0])
print(my_list[-1])

# Also access multiple elements at onece using slicing. Ths syntax for slicing is list[start:stop], start and stop can not be declared but : is required

print(my_list[1:3])
print(my_list[:3])
print(my_list[2:])
print(my_list[:])
```

```python
# 3. Add and remove elements

my_list = [1, 2, 3, 4, 5]
print(my_list)

my_list.append(6) # Insert at the end
print(my_list)
```
