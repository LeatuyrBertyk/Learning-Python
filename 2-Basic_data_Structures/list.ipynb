# List in Python

A list is one of the most commonly used data structures in Python. A list is a collection of items that can hold different types of elements, such as integers, floats, strings, or even other lists. Lists in Python are ordered and mutable, meaning that their elements can be modified after the list is created.

To define a list in Python, you can use square brackets `[]` and seperate the elements with commas.

## Defining Lists and Basic operations

```python
# Defining a simple list
my_list = [1, 2, 3, 4, 5]
print(my_list)

# List with different data types
mixed_list = [1, "Hello", 3.14, True]
print(mixed_list)
```

## Indexing and Slicing in Lists

The index of the first element in a list is 0, the second element is at index 1, so on. Negative indexing starts from the last element with index -1

```python
my_list = [10, 20, 30, 40, 50]
print(my_list[0])
print(my_list[-1])

# Also access multiple elements at onece using slicing. Ths syntax for slicing is list[start:stop], start and stop can not be declared but : is required
print(my_list[1:3])
print(my_list[:3])
print(my_list[2:])
print(my_list[:])
```

## List multiplication

```python
my_list = [1, 2, 3]
duplicated_list = my_list*2

print(duplicated_list)
```

## Insertion and Deletion in Lists

```python
my_list = [1, 2, 3, 4, 5]
print(my_list)

# Add elements
my_list.append(6) # Insert at the end
my_list.insert(1, 7) # Insert at determined position
print(my_list)

# Remove elements
last_element = my_list.pop() # Return and remove the last element
print(last_element)
print(my_list)

my_list.remove(1) # Remove directly value, but make sure this value is in the list
print(my_list)

my_list.pop(2) # Remove by index
print(my_list)


# Delete list
my_list = [1, 2, 3]

del my_list # my_list.clear()
# print(my_list) # Error
```

## Advanced manipulation with List

List comprehensions provide a concise way to create lists. They are more readable and faster than using traditional loops.

### Efficient ways to loop through data with list comprehensions


```python
squares = [x**2 for x in range(1, 11)]
print(squares)

cubes = [x**3 for x in range(1, 11) if x != 3]
print(cubes)
```

### Understanding generators for large-scale data processing

Generators are a type of iterable that generate values on the fly, which makes them memory efficient when dealing with large datasets. You define them using functions and the `yield` keyword.

```python
def square_numbers(n):
    for i in range(n):
        yield i**2

squares_gen = square_numbers(10)

for square in squares_gen:
    print(square)
```

The key advantage of generators is that they do not store all values in memory at once. Instead, they yield one value at a time, making them ideal for large data sets.
