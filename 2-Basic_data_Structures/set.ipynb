# Set in Python

A set is an unordered collection of unique elements. Sets are defined using curly braces `{}`, and they do not allow duplicate values.

## Introduction to Sets

```python
my_set = {1, 2, 3, 4, 5}
print(my_set)
```

## Set operations: union, intersection, difference

```python
set_a = {1, 2, 3}
set_b = {3, 4, 5}

# Union
print(set_a | set_b)

# Intersection
print(set_a & set_b)

# Difference
print(set_a - set_b)
```
