A set is an unordered collection of unique elements. Sets are defined using curly braces `{}`, and they do not allow duplicate values.

```python
# 1. Introduction to Sets

my_set = {1, 2, 3, 4, 5}
print(my_set)
```

```python
# 2. Set operations: union, interseaction, difference

set_a = {1, 2, 3}
set_b = {3, 4, 5}

# Union
print(set_a | set_b)

# Intersection
print(set_a & set_b)

# Difference
print(set_a - set_b)
```
