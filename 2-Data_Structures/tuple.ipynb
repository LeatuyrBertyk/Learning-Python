Tuples are similar to lists in Python, but with one key difference: typles are immutable, meaning that once a tuple is created, its elements cannnot be modified.

Tuples are defined using parentheses `()`.

```python
# 1. Introduction to Tuples: Immutable sequences

my_tuple = (1, 2, 3)
print(my_tuple)

my_tuple = 4, 5, 6
print(my_tuple)
```

```python
# 2. Tuple unpacking and basic operations

# Tuple unpacking allows you to assign the values of ta tuple to multiple variables in one step.


coordinates = (10, 20)
x, y = coordinates

print(x)
print(y)

# Although tuples are immutable, you can perform other operations like indexing and slicing, similar to lists.
my_tuple = (5, 10, 15, 20)
print(my_tuple[1])
print(my_tuple[:3])
```
